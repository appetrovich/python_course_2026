### `else` у цикла: читайте «no break»

Самая неверно понимаемая конструкция языка. `else` у `for` и `while` срабатывает,
**если цикл не был прерван `break`** – а вовсе не «если цикл не выполнился».

Мнемоника: читать не «else», а **«no break»**.

Зачем это нужно: типовая задача «искали и не нашли». Без `else` пришлось бы
заводить флаг `found = False` и проверять его после цикла.

````python
def find_even(numbers):
    for n in numbers:
        if n % 2 == 0:
            print(f"нашли чётное: {n}")
            break
    else:
        # выполнится, только если break не сработал
        print("чётных нет")


find_even([1, 3, 4, 5])   # break сработал -> else пропущен
find_even([1, 3, 5])      # break не сработал -> else выполнился
````

````python
# Без else пришлось бы держать флаг
def find_even_with_flag(numbers):
    found = False
    for n in numbers:
        if n % 2 == 0:
            found = True
            print(f"нашли чётное: {n}")
            break
    if not found:
        print("чётных нет")


find_even_with_flag([1, 3, 5])

# Тот же else есть и у while
n = 0
while n < 3:
    n += 1
else:
    print("while завершился без break, n =", n)
````

### Перехват и выбрасывание исключений
- **`try/except/finally`** – перехват ошибок и действия при завершении.
- **`raise`** – выбросить исключение самому.

О терминах: про `raise` по-русски говорят по-разному – «выбросить исключение»,
«бросить», «возбудить». Мы будем говорить **«выбросить»**; в литературе
и переводах документации встретятся остальные варианты, действие одно и то же.

Для диагностики внутри программы `print` не годится: его нельзя отключить
или отфильтровать. Для этого есть модуль `logging` – разберём его в лекции
про инструменты и качество кода.

````python
# try/except/finally
try:
    x = int("abc")   # ошибка преобразования
except ValueError as e:
    print("Произошла ошибка:", e)
finally:
    print("Этот блок выполняется всегда")

# raise
def divide(a, b):
    if b == 0:
        raise ZeroDivisionError("Деление на ноль запрещено!")
    return a / b

print(divide(10, 2))
print(divide(5, 0))
````

### Исключения подробнее: полная форма `try/except/else/finally`
- **`try`** – защищаемый блок; **`except`** – обработка конкретной ошибки; **`else`** – выполняется, только если ошибок не было; **`finally`** – выполняется всегда (очистка ресурсов).
- Логика разделения: «полезный результат при успехе» кладут в `else`, а не в `try`, чтобы `try` содержал минимум кода и `except` не поймал ошибку случайно из «мирной» части.

````python
def parse_age(text):
    try:
        age = int(text)
    except ValueError as e:            # ловим конкретную ошибку
        print("  except: не число ->", e)
        return None
    else:
        print("  else: ошибок не было")
        return age
    finally:
        print("  finally: выполняется всегда")

print("parse_age('42') ->", parse_age("42"))
print("parse_age('abc') ->", parse_age("abc"))
````

### Несколько типов ошибок и иерархия исключений
- Можно указать несколько блоков `except` подряд – сработает первый подходящий по типу.
- Один `except` может ловить несколько типов через кортеж: `except (ValueError, TypeError)`.
- Исключения образуют **иерархию классов**: `ValueError`, `KeyError`, `ZeroDivisionError` и др. – потомки `Exception`, поэтому `except Exception` перехватит их все (ловить по базовому классу удобно, но делать это надо осознанно).

````python
def lookup_ratio(data, key, divisor):
    try:
        return data[key] / divisor
    except KeyError:
        print("  KeyError: нет такого ключа")
    except ZeroDivisionError:
        print("  ZeroDivisionError: делим на ноль")


d = {"a": 10}
lookup_ratio(d, "b", 2)     # KeyError
lookup_ratio(d, "a", 0)     # ZeroDivisionError

# Один except на несколько типов – через кортеж
try:
    int("x")
except (ValueError, TypeError) as e:
    print("  ValueError или TypeError:", e)

# Ловля по базовому классу: ValueError – подкласс Exception
try:
    raise ValueError("что-то не так")
except Exception as e:
    print(f"  Поймали как {type(e).__name__}: {e}")
````

### Собственные классы исключений
- Свои исключения создаются наследованием от `Exception` (не от `BaseException`).
- Это делает ошибки предметными (`InsufficientFundsError` вместо безликого `ValueError`) и позволяет прикладывать к ним дополнительные данные, доступные в обработчике.

````python
class InsufficientFundsError(Exception):
    """Недостаточно средств на счёте."""
    def __init__(self, balance, amount):
        self.balance = balance
        self.amount = amount
        super().__init__(f"Нельзя снять {amount}: на счёте только {balance}")

def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientFundsError(balance, amount)
    return balance - amount

try:
    withdraw(100, 150)
except InsufficientFundsError as e:
    print("Поймали своё исключение:", e)
    print("Данные ошибки: balance =", e.balance, ", amount =", e.amount)
````

### Цепочки исключений: `raise ... from ...`
- Часто низкоуровневую ошибку оборачивают в доменную, более понятную вызывающему коду.
- `raise NewError(...) from e` сохраняет исходную причину в атрибуте `__cause__` – в трейсбеке будет видно «The above exception was the direct cause…», что бесценно при отладке.

````python
class ConfigError(Exception):
    pass

def load_port(raw):
    try:
        return int(raw)
    except ValueError as e:
        # оборачиваем низкоуровневую ошибку в доменную, не теряя причину
        raise ConfigError(f"Некорректный порт: {raw!r}") from e

try:
    load_port("abc")
except ConfigError as e:
    print("Ошибка конфигурации:", e)
    print("Исходная причина (__cause__):", repr(e.__cause__))
````

### Группы исключений: `ExceptionGroup` и `except*`

Обычный `try` рассчитан на одну ошибку: первая же прерывает выполнение. Но бывают
задачи, где независимых ошибок сразу несколько – проверка формы по многим полям,
разбор пакета файлов, параллельные запросы. Терять все ошибки, кроме первой,
здесь неправильно.

С версии 3.11 такие ошибки собирают в **`ExceptionGroup`** – одно исключение,
внутри которого лежит список остальных.

Разбирают группу звёздочкой: `except*` забирает из группы **все** подходящие
исключения, а остальные оставляет в ней. Поэтому сработать может несколько веток
`except*` подряд – в обычном `try` такое невозможно.

Смешивать `except` и `except*` в одном `try` нельзя: это синтаксическая ошибка.

````python
def validate(form):
    errors = []
    if not form.get("name"):
        errors.append(ValueError("имя не заполнено"))
    if "@" not in form.get("email", ""):
        errors.append(ValueError("почта без @"))
    if form.get("age", 0) < 18:
        errors.append(PermissionError("нет 18 лет"))
    if errors:
        raise ExceptionGroup("форма заполнена неверно", errors)


try:
    validate({"name": "", "email": "ivan", "age": 15})
except* ValueError as group:                 # сработают обе ветки, а не первая
    print("ошибки заполнения:", [str(e) for e in group.exceptions])
except* PermissionError as group:
    print("ошибки доступа:", [str(e) for e in group.exceptions])
````

### Как `except*` разбирает группу

Обычный `except` смотрит на исключение целиком: `ExceptionGroup` для него – объект
одного типа, поэтому `except ValueError` его не поймает, даже если внутри лежит
`ValueError`. Это первый пример в ячейке.

`except*` устроен иначе. Он **делит** группу на две части: подходящие исключения
уходят в обработчик, остальные остаются в ней. Когда все ветки `except*` отработали,
интерпретатор собирает необработанный остаток в новую группу и выбрасывает её
дальше – наружу, как обычное исключение.

Во втором примере поэтому и получается, что `ValueError` обработан внутри,
а `TypeError` долетает до внешнего `try`. Если бы веток `except*` хватило на все
исключения, наружу не ушло бы ничего.

````python
# Обычный except ловит саму группу целиком, а не то, что внутри неё
try:
    raise ExceptionGroup("две беды", [ValueError("раз"), TypeError("два")])
except ValueError:
    print("сюда не попадём")
except ExceptionGroup as group:
    print("поймали группу:", [type(e).__name__ for e in group.exceptions])

# То, что не разобрали через except*, остаётся в группе и уходит дальше
try:
    try:
        raise ExceptionGroup("две беды", [ValueError("раз"), TypeError("два")])
    except* ValueError as group:
        print("обработали:", [str(e) for e in group.exceptions])
except ExceptionGroup as rest:
    print("осталось необработанным:", [type(e).__name__ for e in rest.exceptions])
````

### Синтаксис: двоеточие и отступы

Условная инструкция в Python короче, чем в C-подобных языках: не нужны ни круглые
скобки вокруг условия, ни фигурные скобки вокруг блока, ни точки с запятой в конце
строк. Вместо них два правила:

- **конец строки = конец инструкции**;
- **отступ = границы блока**, а заголовок составной инструкции заканчивается
  двоеточием `:`.

Двоеточие и отступ работают одинаково во всех составных инструкциях: `if`, `for`,
`while`, `def`, `class`, `try`, `with`.

Отступы – часть синтаксиса, а не оформления. Там, где границы блока задают скобки,
отступы каждый расставляет по-своему и визуальная структура расходится с фактической.
В Python разойтись они не могут: как код выглядит, так он и работает. Смешивать
табы и пробелы внутри одного блока нельзя.

````python
# Так могло бы выглядеть в C-подобном языке:
# if (x > y) {
#     x = 1;
#     y = 2;
# }

x, y = 5, 3
if x > y:
    x = 1
    y = 2
print(x, y)  # 1 2
````

### Несколько особых случаев
- По умолчанию:  
  - конец строки = конец инструкции  
  - отступ = границы блока  
- Есть специальные исключения:  
  1. **Несколько инструкций в одной строке** через `;`  
  2. **Инструкция на несколько строк** – в скобках `()`, `[]`, `{}`  
  3. **Старый способ** – обратный слэш `\` (лучше не использовать)  
  4. **Однострочные блоки**: `if x > y: print(x)` (только для простых инструкций)  
- Всё это допустимо, но считается не самым читаемым стилем  


````python
# 1. Несколько инструкций на одной строке
a = 1; b = 2; print(a + b)

# 2. Продолжение в скобках
mylist = [
    1111, 2222,
    3333, 4444
]

X = (1 + 2 +
     3 + 4)

# 3. Обратный слэш (устаревший способ, лучше не использовать)
Y = 10 + 20 + \
    30 + 40

# 4. Однострочный if
if X > Y: print("X больше Y")
````

## Присваивания, выражения и вывод
- **Присваивание (=)**: связывает имя с объектом (создаёт ссылку)  
- Переменные (имена) создаются при первом присваивании  
- Нельзя использовать имя до присваивания – будет ошибка  
- Присваивание может происходить не только явно, но и неявно (например, в `for`, `def`, `import`)  


````python
# Простые примеры присваивания
x = 10          # создаём имя x и связываем с числом 10
y = x + 5       # имя y ссылается на результат выражения

# Ошибка: обращение к имени до присваивания
# print(z)   # NameError: name 'z' is not defined

# Неявные присваивания
for i in range(3):      # i создаётся автоматически
    print(i)

def func(a):            # аргумент a присваивается при вызове
    return a * 2
print(func(5))
````

### Формы присваивания в Python
- **Обычное присваивание**: `x = obj`  
- **Кортежное и списковое присваивание** (unpacking): `a, b = (1, 2)`  
- **Присваивание последовательностей**: `a, b, c = "abc"`  
- **Расширенная распаковка**: `a, *b = "hack"`  
- **Множественное присваивание**: `x = y = 10` (оба имени ссылаются на один объект)  
- **Составные присваивания** (augmented): `x += 1`, `y *= 2`  
- **Оператор := (морж)** – присваивание внутри выражения  


### Базовые и последовательностные присваивания
- **Базовое присваивание**: имя, индекс, срез или атрибут слева = выражение справа  
- Примеры:  
  - имя: `x = 10`  
  - индекс: `L[0] = 3`  
  - срез: `L[-1:] = [4, 5]`  
  - атрибут: `obj.attr = value`  
- **Распаковка (unpacking)**: кортежи и списки слева = итерируемый объект справа  
- Значения сопоставляются по позициям слева направо  
- Правая часть вычисляется **целиком** до присваивания и складывается
  во временный кортеж – только потом он распаковывается по целям слева  
- Отсюда обмен значениями без временной переменной: `a, b = b, a`  
- На правой стороне может быть любой итерируемый объект, не только список или кортеж  


````python
# Базовое присваивание
L = [1, 2]
L[0] = 3        
L[-1:] = [4, 5] 
print(L)        

# Атрибут 
class Obj: pass
o = Obj()
o.attr = L
print(o.attr)
````

````python
# Распаковка (tuple / list assignment)
first, second = 1, 2
print(first, second)   

[C, D] = [3, 4]
print(C, D)            
````

````python
# Обмен значениями
a, b = 10, 20
a, b = b, a
print(a, b)

# Справа сначала собирается кортеж, и уже он распаковывается в цели слева
a, b = 0, 1
a, b = b, a + b
print(a, b)            # 1 1 - обе формулы посчитаны на старых значениях

# Если бы присваивание шло по очереди, вышло бы другое
a, b = 0, 1
a = b
b = a + b
print(a, b)            # 1 2
````

````python
# Распаковка с разными типами
[a, b, c] = (1, 2, 3)
print(a, c)                     
````

````python
(a, b, c) = "ABC"
print(a, c)   
````

### Расширенные шаблоны присваивания последовательностей
- Обычно количество целей слева и значений справа должно совпадать  
- Если их разное количество → `ValueError`  
- Решение: использовать срезы (`string[:2]`, `string[2:]`)  
- Можно комбинировать индексы и срезы, чтобы «собрать» нужное распределение  
- Поддерживаются **вложенные последовательности**: `(a, b), c = ('TE', 'XT')`  
- Частый приём: распаковка `range()` → инициализация нескольких переменных сразу  
- В циклах можно разделять элементы на «первый и остальные» (`front, L = L[0], L[1:]`)  


````python
# Стандартное правило: длины должны совпадать
string = "TEXT"
a, b, c, d = string
print(a, b, c, d)   # T E X T
````

````python
# Ошибка: несоответствие длин
a, b, c = string  
````

````python
# Обходное решение через срезы
a, b, c = string[0], string[1], string[2:]
print(a, b, c)      # T E XT
````

````python
# Вложенные присваивания
((a, b), c) = ("TE", "XT")
print(a, b, c)      # T E XT
````

````python
# Инициализация переменных через range
red, green, blue = range(3)
print(red, blue)    # 0 2
````

````python
# Деление списка на первый элемент и "хвост"
L = [1, 2, 3, 4]
while L:
    front, L = L[0], L[1:]
    print(front, L)
````

### Расширенная распаковка со звёздочкой
- Обычно количество целей слева = количество элементов справа, иначе `ValueError`  
- **Звёздочка `*`** забирает «остаток» элементов и **всегда даёт список**, даже если
  элемент один; если не осталось ничего – пустой список `[]`  
- `*` может стоять в начале, в конце или в середине слева от `=`  
- Работает с любыми итерируемыми: списками, строками, `range` и другими  
- В отличие от срезов, `*` возвращает список, а не исходный тип  

**Ошибки:**
- более одной звёздочной переменной в одном блоке  
- слишком мало целей слева без `*`  
- звёздочка вне последовательности (`*a = ...`)  

**Где ещё работает:** в `for` переменная цикла – обычная цель присваивания, поэтому
распаковка со `*` работает и там. Символы `*` и `**` применяются также в объявлениях
функций, вызовах и литералах.

Типовые шаблоны: «первый и остальные», «все кроме последнего».


````python
seq = [1, 2, 3, 4]

# С * – можно забирать остаток
a, *b = seq
print(a)
print(b)
````

````python
# Звёздочка в начале
*a, b = seq
print(a, b)   
````

````python
# Звёздочка в середине
a, *b, c = seq
print(a, b, c)  
````

````python
# Работает и со строками, и с range
a, *b = "hack"
print(a, b)     

a, *b, c = range(4)
print(a, b, c)  
````

````python
# Упрощение "первый и остальные"
L = [1, 2, 3, 4]
while L:
    front, *L = L
    print(front, L)
````

````python
# Ничего не осталось для *
a, b, c, d, *e = seq
print(a, b, c, d, e)  # 1 2 3 4 []
````

````python
# * только один раз
a, *b, c, *d = seq   
````

````python
# * вне последовательности
*a = seq             
````

````python
*a, = seq              
a
````

````python
# Расширенная распаковка в цикле for
for a, *b, c in [(1, 2, 3, 4), (5, 6, 7, 8)]:
    print(a, b, c)
````

````python
# Эквивалент через срезы (менее удобно)
for seq in [(1, 2, 3, 4), (5, 6, 7, 8)]:
    a, b, c = seq[0], seq[1:-1], seq[-1]
    print(a, b, c)
````

### Множественное присваивание
- Запись вида `a = b = c = значение` присваивает одно и то же значение нескольким переменным  
- Все переменные начинают ссылаться на **один и тот же объект**  
- Для неизменяемых объектов (числа, строки, кортежи) это безопасно  
- Для изменяемых объектов (списки, словари) это может привести к неожиданным эффектам  
- Чтобы избежать проблем, создавайте новые объекты отдельно: `a = []; b = []` или `a, b = [], []`  


````python
# Несколько переменных указывают на один и тот же объект
a = b = c = "code"
print(a, b, c)   
````

````python
# Для неизменяемых объектов (чисел) это безопасно
a = b = 0
b = b + 1
print(a, b)      
````

````python
# Но с изменяемыми объектами нужно быть осторожным
a = b = []
b.append(42)
print(a, b)      
````

````python
# Решение: создавать новые объекты
a = []
b = []
b.append(42)
print(a, b)      
````

````python
# То же самое в одной строке
a, b = [], []
print(a, b)      
````

### Сокращённые присваивания 
- Запись вида `X += Y` – это сокращение для `X = X + Y`  
- Работают для любых типов, где определены соответствующие операции  
- Примеры: `+=`, `-=`, `*=`, `/=`, `//=`, `%=` и др.  
- Преимущества:  
  - Меньше кода для написания  
  - Левая часть вычисляется один раз (быстрее)  
  - Для изменяемых объектов применяются оптимизации (операции в месте, без создания копий)  
- Для неизменяемых типов (`int`, `str`, `tuple`) результат будет новым объектом  
- Для изменяемых типов (`list`, `dict`, `set`) операция часто меняет объект на месте  


````python
# Пример с числами
x = 1
x = x + 1   # обычная форма
print(x)    
x += 1      # сокращённая форма
print(x)    
````

````python
# Пример со строками (конкатенация)
s = "hack"
s += "HACK"
print(s)    
````

````python
# Пример со списком (in-place extend)
L = [1, 2]
L += [3, 4]
print(L)    
````

````python
# Важно: += и + могут вести себя по-разному
L = []
L += "hack"
print(L)   
````

````python
L = L + "hack"
````

````python
# Разница в поведении при общих ссылках
L = [1, 2]
M = L
L = L + [3, 4]   # создаётся новый список
print(L, M)     
````

````python
L = [1, 2]
M = L
L += [3, 4]      # изменяется исходный список
print(L, M)      
````

### Именованные выражения присваивания (оператор `:=`)
- Появились в Python 3.8  
- Запись `имя := выражение` одновременно:  
  - присваивает значение переменной  
  - возвращает это значение как результат выражения  
- Можно использовать в местах, где обычное `=` недопустимо (например, в заголовке `if` или `while`)  
- Ограничения:  
  - слева может быть только имя переменной  
  - не поддерживает формы вроде `:+=`  
- Использовать стоит умеренно: удобнее, но иногда ухудшает читаемость  


````python
# Пример: присваивание и использование в одном выражении
a = "hack" * (b := 2)
print(a, b)   
````

````python
# Использование в условиях
lines = ["one\n", "two\n", ""]
it = iter(lines)

while (line := next(it)) != "":
    print(line.strip())
````

````python
# В if
ignore = "skip\n"
line = "skip\n"
if (line := line) != ignore:
    print(line)   
````

### Инструкция `match`
- Появилась в Python 3.10.  
- Используется для множественного выбора (аналог `switch` в других языках).  
- Может заменять цепочки `if/elif/else` или словари для выбора действий.  
- Синтаксис: `match <выражение>:` и далее `case <значение>:`.  
- `case _:` – это «по умолчанию», срабатывает, если нет совпадений.  
- Несколько значений перечисляются через `|`, а `as` связывает совпавшее значение
  с переменной – этого у `if` нет.  

**Когда что брать:** `match` выигрывает на множественном выборе по значению и
делает разбор структур явным. `if` гибче: в нём можно написать произвольно
сложное условие.


````python
# Пример через if/elif/else
state = "go"
if state == "go":
    print("green")
elif state == "stop":
    print("red")
else:
    print("yellow")
````

````python
# Пример через словарь
colors = {"go": "green", "stop": "red"}
print(colors.get(state, "yellow"))
````

````python
# Пример через match
match state:
    case "go":
        print("green")
    case "stop":
        print("red")
    case _:
        print("yellow")
````

````python
# Использование | 
state = "halt"
match state:
    case "go" | "proceed" | "start":
        print("green")
    case "stop" | "halt" as what:
        print("red")
        print("means", what)
    case other:
        print("catchall:", other)
````

### Продвинутый `match` в Python
- Расширяет простую множественную проверку до **структурного сопоставления с образцом**.  
- Поддерживает разные виды шаблонов:  
  - **Literal** – сравнение по значению (`case 1 | 2`).  
  - **Wildcard** – `_` совпадает с чем угодно.  
  - **Capture** – имя переменной сохраняет совпавшее значение.  
  - **Or** – несколько вариантов (`case 1 | 2 | 3`).  
  - **As** – связывает совпавшее значение с переменной.  
  - **Sequence/Mapping/Instance patterns** – можно проверять списки, словари и объекты по структуре.  
- Можно использовать `if`-условия в заголовке `case` (guards).  


````python
# Пример структурного сопоставления
state = [1, 2, 3]

match state:
    case 1 | 2 | 3 as what:
        print("or", what)                # совпадение с литералом
    case [1, 2, what]:
        print("list", what)              # совпадение со списком
    case {"a": 1, "b": 2, "c": what}:
        print("dict", what)              # совпадение с dict
    case (1, 2, what):
        print("tuple", what)             # совпадение с кортежем
    case _ as other:
        print("other", other)            # совпадение со всем остальным
````

## Частые ошибки

- **Голый `except:` (или `except Exception`) на всякий случай.** Он проглатывает и опечатки,
  и `KeyboardInterrupt` – ошибка молча превращается в неверный результат. Ловите конкретный тип.
- **`return` внутри `finally`.** Он затирает и результат, и исключение из `try` –
  ошибка исчезает бесследно.
- **Повторный проход по генератору.** Генератор одноразовый: второй цикл увидит пустоту.
  Нужен повторный проход – материализуйте в `list(...)`.
- **`raise` без контекста.** Выбрасывая новое исключение поверх пойманного, пишите `raise NewError(...) from e`,
  иначе исходная причина теряется в трассировке.


````python
data = {"age": "30"}

# ПЛОХО: except без типа проглатывает ЛЮБУЮ ошибку, включая опечатку
try:
    age = int(data["aeg"])           # опечатка в ключе -> KeyError
except:
    age = 0
print("ошибку проглотили, age =", age)

# ХОРОШО: ловим только то, что ожидали
try:
    age = int(data["aeg"])
except ValueError:                   # KeyError сюда НЕ попадёт
    age = 0
except KeyError as e:
    print("проблема видна сразу:", type(e).__name__, e)
````

### 1. Генератор `chunks`

Напишите генератор `chunks(seq, n)`, который отдаёт элементы `seq` кусками по `n`
(последний кусок может быть короче). Затем убедитесь, что генератор **одноразовый**:
повторный проход по тому же объекту не даст ничего.

```python
list(chunks(range(7), 3))   # [[0, 1, 2], [3, 4, 5], [6]]
```


````python
def chunks(seq, n):
    """Генератор: режет seq на куски по n элементов."""
    buf = []                      # накопитель текущего куска
    for item in seq:
        buf.append(item)
        if len(buf) == n:         # набрали ровно n - отдаём кусок наружу
            yield buf
            buf = []              # новый список; buf.clear() испортил бы уже отданный
    if buf:                       # то, что осталось в накопителе, короче n
        yield buf


gen = chunks(range(7), 3)
print("куски:", list(gen))
print("второй проход:", list(gen))   # генератор одноразовый
````

### 2. Своё исключение и `raise ... from`

Напишите `parse_age(text)`: возвращает целое число, а при неверном вводе выбрасывает
собственное исключение `AgeError`. Требования:

- нечисловая строка → `AgeError`, но **исходный** `ValueError` сохранён как причина
  (`raise ... from e`, доступно через `e.__cause__`);
- число вне диапазона 0…120 → `AgeError` без причины.


````python
class AgeError(ValueError):
    """Ошибка предметной области: возраст не распознан."""


def parse_age(text):
    try:
        age = int(text)
    except ValueError as e:
        raise AgeError(f"не возраст: {text!r}") from e
    if not 0 <= age <= 120:
        raise AgeError(f"вне диапазона: {age}")
    return age


for value in ("30", "тридцать", "500"):
    try:
        print(value, "->", parse_age(value))
    except AgeError as e:
        cause = type(e.__cause__).__name__ if e.__cause__ else "нет"
        print(value, "-> AgeError:", e, "| причина:", cause)
````

### 3. `match`

Напишите `run(command)`, разбирающий команды-списки через `match`:

- `["go", "восток"]` → `"идём на восток"`;
- `["take", "меч", 3]` → `"берём 3 × меч"`;
- `["quit"]` и `["exit"]` → `"выход"`;
- всё остальное → `"не понял команду"`.

Подсказка: шаблон `case ["take", item, int() as count]:` заодно проверит тип.


````python
def run(command):
    match command:
        case ["go", direction]:
            return f"идём на {direction}"
        case ["take", item, int() as count]:       # int() заодно проверяет тип
            return f"берём {count} × {item}"
        case ["quit"] | ["exit"]:
            return "выход"
        case _:
            return "не понял команду"


for cmd in (["go", "восток"], ["take", "меч", 3], ["quit"], ["take", "меч", "три"]):
    print(cmd, "->", run(cmd))
````
