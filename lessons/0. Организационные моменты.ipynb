## Рабочее окружение: venv, pip и git

Прежде чем писать код и сдавать домашки, нужно разобраться с тремя вещами:
**установить сам Python**, **изолировать зависимости проекта** (venv + pip)
и **вести историю кода** (git). Это базовая профессиональная гигиена: без неё
сдача заданий превращается в отдельную проблему.


## Установка Python

Курс идёт на **Python 3.13**, код рассчитан на **3.11 и новее**. На более старых
версиях часть примеров не запустится: `match` и запись `str | None` появились
в 3.10, группы исключений – в 3.11.

| ОС | Что делать | Где обычно ломается |
|---|---|---|
| **Windows** | установщик с [python.org](https://www.python.org/downloads/), при установке отметить **«Add python.exe to PATH»** и оставить включённым «tcl/tk and IDLE» | без галочки PATH команда `python` не находится в терминале. Версию из Microsoft Store лучше не ставить: она живёт в изолированной папке и путается с обычной |
| **macOS** | установщик с [python.org](https://www.python.org/downloads/) либо пакетный менеджер, если вы им уже пользуетесь | системный Python трогать нельзя, на нём работает сама ОС. Команда всегда `python3`, не `python` |
| **Linux** | пакетный менеджер дистрибутива, например `sudo apt install python3 python3-venv python3-pip` | `python3-venv` и `python3-pip` ставятся **отдельными пакетами**; без первого не сработает `python3 -m venv` |


## Проверка и несколько версий на одной машине

Сразу после установки проверьте две вещи: интерпретатор находится и модуль `venv` на месте.

```bash
# Windows
python --version
python -m venv --help
where python

# macOS и Linux
python3 --version
python3 -m venv --help
which python3
```

**Если версий несколько.** Это нормальная ситуация: система принесла свою,
вы поставили свою, а какая-то программа притащила третью.

```bash
# Windows: лаунчер py показывает и выбирает версию
py --list
py -3.13 -m venv .venv

# macOS и Linux: версия указывается прямо в имени команды
python3.13 -m venv .venv
```

Окружение запоминает тот интерпретатор, которым было создано, поэтому выбирать
версию нужно один раз – в момент создания `.venv`.


## Виртуальное окружение (venv)

**Проблема:** если ставить пакеты «глобально», проекты начинают конфликтовать
(одному нужен `numpy 1.x`, другому `2.x`). **Решение:** у каждого проекта –
своё изолированное окружение.

```bash
# 1. Создать виртуальное окружение в папке .venv
python -m venv .venv        # Windows
python3 -m venv .venv       # macOS и Linux

# 2. Активировать его
.venv\Scripts\activate       # Windows
source .venv/bin/activate   # macOS и Linux

# После активации в начале строки терминала появляется (.venv)

# 3. Выйти из окружения, когда закончили
deactivate
```


## Установка пакетов (pip)

`pip` – стандартный менеджер пакетов Python. Ставить пакеты нужно **при активированном** `.venv`.

```bash
pip install numpy matplotlib      # установить пакеты
pip list                          # посмотреть, что установлено
pip freeze > requirements.txt     # зафиксировать версии в файл
pip install -r requirements.txt   # воспроизвести окружение по файлу
```

**`requirements.txt`** – текстовый список зависимостей проекта. Его коммитят в git,
чтобы любой мог одной командой развернуть то же окружение.

Существуют и более современные инструменты – `uv`, `poetry`: они совмещают установку
пакетов с управлением версиями самого Python (`uv python install 3.13`).
Им посвящён отдельный доклад.


## Система контроля версий (git): зачем

**git** запоминает историю изменений кода: можно откатиться к любой прошлой версии, вести параллельные ветки, совместно работать над проектом и – в нашем случае – **сдавать домашки**.

Ключевые понятия:
- **repository (репозиторий)** – папка проекта под контролем git;
- **commit** – снимок состояния проекта в момент времени с описанием;
- **branch (ветка)** – независимая линия разработки;
- **remote** – удалённая копия репозитория (например, на GitHub).

## git: минимальный рабочий цикл

```bash
git init                        # создать репозиторий в текущей папке
git status                      # что изменилось с последнего коммита

git add hw1.py                  # добавить файл в индекс (staging)
git add .                       # добавить все изменения
git commit -m "Домашка 1: решение"   # зафиксировать снимок

git log --oneline               # краткая история коммитов
```

**`.gitignore`** – файл со списком того, что git должен игнорировать (не коммитить): `.venv`, `__pycache__`, `.env` и т. п. Без него в репозиторий попадёт мусор и секреты.

## git + GitHub

```bash
# Связать локальный репозиторий с удалённым (один раз)
git remote add origin https://github.com/USER/REPO.git

git push -u origin main         # отправить коммиты на GitHub (первый раз)
git push                        # последующие отправки
git clone https://github.com/USER/REPO.git   # скачать чужой/свой репозиторий
```

GitHub хранит удалённую копию репозитория: она переживёт потерю ноутбука, её видно
другим людям, и именно через неё удобно показывать свою работу. Как через это будут
сдаваться домашки, разберём отдельно.


## Грабли первого дня

**git не знает, кто вы.** Первый же `git commit` упадёт с `Please tell me who you are`.
Настраивается один раз на машину:

```bash
git config --global user.name "Иван Иванов"
git config --global user.email "ivan@example.com"
```

**Ветка называется не так.** Старый git создаёт `master`, а пушим мы в `main`:

```bash
git init -b main      # при создании репозитория
git branch -M main    # переименовать, если уже создан
```

**Windows не даёт активировать окружение.** `.venv\Scripts\Activate.ps1` блокируется
политикой PowerShell – разрешается один раз:

```powershell
Set-ExecutionPolicy -Scope CurrentUser -ExecutionPolicy RemoteSigned
```

**`python` – не тот, что вы думаете.** `python`, `python3` и интерпретатор из `.venv` –
часто разные программы. Признак активированного окружения – `(.venv)` в начале строки
терминала; проверить можно через `python --version` и `which python`
(в Windows – `where python`): путь должен вести в `.venv`. Без активации пакеты
ставятся «глобально», и проекты начинают конфликтовать.

**И две привычки, которые стоит завести сразу:**
- не коммитить `.venv/` и файлы `.env` с паролями – для этого нужен `.gitignore`;
- не делать один коммит «final» в конце: коммитьте небольшими осмысленными шагами.
